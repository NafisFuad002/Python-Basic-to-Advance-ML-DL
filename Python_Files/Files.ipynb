```python
# akta variable er moddhe file ta open kora holo
a = open("./sample_data/231311058.txt","r")
# "r" holo read mode, shudhu file e ki ase porbo
# "w" holo write mode, write korle ager sob data del hye jabe then new data add hobe
# "a" holo append mode, ager data thakbe, new data add hobe

b = a.read() # file e ja ja ase seta ei variable e rakhbo

print(b) # print korle kisu show hobe na file e kisu nai
```

```python
# ei close er kaj ta amra with keyword use kore korte pari tahole auto close hye jabe
with open("./sample_data/231311058_2.txt","r") as variable_v:
  variable_r = variable_v.read()
  print(variable_r)

print(variable_v.closed)
```

```python
# file na thakle auto new file khula hobe
li=["hello", "how are you?", "go home"]

with open("./sample_data/231311058_3.txt","a") as m: # new file
  m.writelines(li)

with open("./sample_data/231311058_3.txt","r") as n:
  print(n.read())
```

```python
# file pointer details
with open("./sample_data/231311058_3.txt") as aa:
  print(aa.tell()) # file pointer kothai ase ta show korbe
  print(aa.read(5)) # 5 number index porjonto print hobe
  print(aa.tell())
  print(aa.read())
  print(aa.tell())
  aa.seek(10)       # cursor k 10 number index e niye jabe
  print(aa.tell())
  print(aa.read())
  aa.seek(0)        # cursor k 0 number index e niye jabe
  print(aa.read())
```

```python
# "w+" mode e read write 2 tai kora jai
# write append er moto kaj kore
with open("./sample_data/231311058_4.txt","w+") as q:
  q.write("hello\n")
  q.seek(0)
  print("Print-1: ",end="")
  print(q.read())
  q.write("how are you?\n")
  q.seek(0)
  print("Print-2: ",end="")
  print(q.read())
  q.truncate(6)
  q.seek(0)
  print("Print-3: ",end="")
  print(q.read())
```
