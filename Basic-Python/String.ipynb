```python
s3 = s1[0:7]
s4 = s2[0:16:2]
s5 = s1[-1] # -1 means last index
s6 = s1[-2] # -2 means before last index
s7 = s1[-3]
s8 = s1[-12]
print(s3)
print(s4)
print(s5)
print(s6)
print(s7)
print(s8)
```

**String Methods (Most used)**

```python
s9 = "Hello World"
s10 = "Welcome back to python"

print(len(s9))
print("back" in s10)
print("Back" in s10)

s11 = s10.lower()
s12 = s11.upper()
s13 = s11.capitalize()

print(s11)
print(s12)
print(s13)

print(s9.find("to"))
print(s10.find("to")) # 2ta "to" thakle 1st er tar index return korbe

s14 = "to to top"

print(s14.find("to"))
print(s14.rfind("to")) # rfind() use korle ses theke khujbe

print(s14.find("to "))
print(s14.rfind("to "))
```

```python
s15 = "hi hi hihi hi"
s16 = "hi hi hihi hi "
s17 = s16.replace("hi","o")

print(s15.count("hi"))
print(s15.count("hi "))
print(s16.count("hi "))
print(s17)
```

```python
s18 = "to to to top"

print(s18.endswith("op"))
print(s18.endswith("top"))
print(s18.startswith("to"))
print(s18.startswith("to "))
print(s18.startswith("to  "))
```

```python
s19 = "I eat rice"
s20 = s19.split()
s21 = "I-eat-rice"
s22 = s21.split("-")

print(s19)
print(s20)
print(s21)
print(s22)
```

```python
# Taking multiple word input for cp
inp = input("Enter a sentence: ")
after_splitting = inp.split()
print(after_splitting)
```

```python
from re import U
# Joining
s23 = ['I', 'go', 'to', 'school']
s24 = " ".join(s23)
s25 = "-".join(s23)
s26 = "_".join(s23)

print(s24)
print(s25)
print(s26)

print()

print(type(s23))
print(type(s24))
```

```python
# Formated String
name = "Magician"
age = 30
country = "Bangladesh"
cg = 3.8345634573458

info = f"My name is {name.upper()}. I'm {age} years old. \nI live in {country}. My CGPA is {cg:.3} out of 4."

print(f"My name is {name.upper()}. I'm {age} years old. \nI live in {country}. My CGPA is {cg:.3} out of 4.\n")
print(info)
```
