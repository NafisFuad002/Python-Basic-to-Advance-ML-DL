```python
li1 = [10,20,30,40.5,"Hello"]

print(type(li1))
print(li1)
print(li1[2])
print()

li1[1] = li1[0] + li1[2]
li2 = li1[2:5]

print(li1)
print(li2)
print()

li2.append("World") # last e add kore
li1.insert(0,"World") # fixed index e add kore

print(li2)
print(li1)
```

```python
li3 = [1,2,3,4,5,6,7]
li3.pop() # last element del kore

print(li3)
print(li3.pop())
print(li3)
```

```python
# 2D list
li5 = [[10, 20], [5.5, 6.4, 9.9], ['Hello', 'World']]

print(li5)
print(li5[2])
print(li5[1][2])
```

```python
li6 = [5,2,8,4,10,1,7.3]
li7 = li6.sort() # sort() none return kore
print(li7)
print(li6)
print()

li8 = li6.sort(reverse=True)
print(li8)
print(li6)
print()

li9 = [1,6,8,2,3]
li9.reverse()

print(li9)
```

```python
# Stack using list (Last in First out)
stack = []
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)

print(stack)
print(f"Length of stack = {len(stack)}")

print(stack.pop())
print(stack)
print(stack.pop())
print(stack)
print(f"Length of stack = {len(stack)}")
print(f"Top element = {stack[-1]}")
```

```python
# Queue using list (First in First out)
queue = []
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)

print(queue)
print(f"Length of queue = {len(queue)}")

print(queue.pop(0))
print(queue)
print(queue.pop(0))
print(queue)
print(f"Length of stack = {len(queue)}")
print(f"Top element = {queue[-1]}")
```

```python
# List Comprehension
li10 = [i for i in range(10)]
li11 = [i*2 for i in range(10)]
li12 = [i for i in range(10) if i%2==0 and i>0]

print(li10)
print(li11)
print(li12)
```

```python
li13 = ["i", "go", "to", "school"]
li14 = [i.capitalize() for i in li13 if i!="school"]

print(li14)
```

```python
li15 = [1,2,4.4,"hi"]
li16 = li15[::-1]

print(li16)
```
