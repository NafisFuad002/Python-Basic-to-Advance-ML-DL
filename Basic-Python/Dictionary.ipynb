```python
# Dictionary stores data with keys
# it is mutable and unordered. So, can't use indexing
# Key is unmutable and unique

a = {}
b = {"hello"}
c = {"id" : 41,"name" : "Done", "marks" : [90,95,92]}

print(type(a))
print(a)
print()
print(type(b))
print(b)
print()
print(type(c))
print(c)
```

```python
d = {1 : "hello", "num" : [10,20]}

print(d["num"]) # key na thakle error dekhabe
print(d.get("num")) # key na thakle 'None' dekhabe

d[1] = "world"
print(d)
```

```python
e = {"name" : "abc", "id" : 58, "phn" : 99, "id" : 54}

print(e)

e["grade"] = "A+"

print(e)

e.update({"dept" : "CSE"})

print(e)

del e["name"]

print(e)
```

```python
f = {"id" : 23, "grade" : "A+"}
g = f.copy()

print(g)

# h = {[10,20] : "marks"} -> error
h = {(10,20) : "marks"} # not error because tauple is unmutable

print(h)
```

**View methods**

```python
i = {"id" : 58, "name" : "hello", "cga" : 4.00}
variable_a = i.keys()
variable_b = i.values()
variable_c = i.items()

print(variable_a)
print(variable_b)
print(variable_c)
print()

i["location"] = "Dhaka" # keys(),values(),items() real time e kaj kore

print(variable_a) # variable_a update kora hy nai, auto update hye jabe
print(variable_b) # variable_b update kora hy nai, auto update hye jabe
print(variable_c) # variable_c update kora hy nai, auto update hye jabe
```

**Dictionary Comprehension**

```python
for variable_a,variable_b in i.items():
    print(variable_a,variable_b)
```

```python
j = {x:x**2 for x in range(5)}
print(j)

k = {x:x*2 for x in range(5) if x%2==0}
print(k)
```

```python
keys = [231311050, 231311060, 231311134]
values = ["Shakil", "Shovon", "Shahriar"]

l = {x:y for x,y in zip(keys, values)}

print(type(l))
print(l)
```
