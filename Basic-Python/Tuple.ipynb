```python
# Tuple is immutable (Change kora jai na)

a = (10,20.5,"Hello")
b = (50) # 1ta value thakle seta tuple hy na
c = (50,)
d = tuple([50])
e = ([40,42])

print(type(a))
print(a)
print()

print(type(b))
print(b)
print()

print(type(c))
print(c)
print()

print(type(d))
print(d)
print()

print(type(e))
print(e)
print()
```

```python
list_1 = [10,11,12]
f = (22,23,24)
g = tuple(list_1)
h = list(f)

print(type(g))
print(g)
print()
print(type(h))
print(h)
```

```python
# Access
i = (1,2,3,4,5,6,7)
j = i[1:3]

print(i[2])
print(j)
```
