```python
# Set is unique, mutable, unordered
# We can not use indexing and slicing technique because it is unordered

a = {10,40,30,20,10,40,40,40,5.5,"Hello",10,"Hello"}

print(type(a))
print(a)
```

```python
g = {10,20,30,20,30,20,20,10,50}

if 10 in g:
  print("Present")
else:
  print("Absent")

for i in g:
  print(i, end=" ")
print()

g.add("Hello")

print(g)
```
