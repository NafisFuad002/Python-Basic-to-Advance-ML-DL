```python
#Basic structure of a function
def fun():
  print("Hello World!")

fun()
```

```python
#Parameterized function
def fun2(x): # 'x' is parameter
  while x>0:
    x=x-1
    print("Hello World!")

fun2(3) # '3' is argument
```

```python
#Default Parameter
def fun3(x="Python"):
  print(f"I am {x}.")

fun3()
fun3("Java")
```

```python
#Return value
def fun4(x,y):
  sum=x+y
  return sum

ans = fun4(54,58)
print(ans+5)
```

```python
#Doc
def fun5(x):
  """
  This function prints the square of a number.
  """
  print(x*x)

fun5(5)

help(fun5) # ami j description ta likhsi seta dekhabe
```

```python
#Positional arguments
def fun6(*args):
  print(args)

fun6(10,40,30,20) #serially tuple hisabe chole jabe
```

```python
#Keyword arguments
def fun7(a,b,c):
  print(a,b,c)

fun7(c=90,a=30,b=60)
```

```python
#Variable Side Keyword arguments
def fun6(**kwargs):
  print(kwargs)

fun6(a=10,c=40,b=30,d=20) #serially dictionary hisabe chole jabe
```

```python
#Return more than one value
def fun8(a,b,c):
  a=a*2
  b=b*2
  c=c*2
  return a,b,c

x,y,z = fun8(5,6,7)
m = fun8(2,3,4) # tuple hisabe asbe

print(x,y,z)
print(m)
```

```python
#Return more than one value as list
def fun9(a,b,c):
  a=a*2
  b=b*2
  c=c*2
  return [a,b,c]

m = fun9(2,3,4) # list hisabe asbe
print(m)
```

```python
#generator
variable_a=[x for x in range(500)]

def fun10(x, variable_a):
  for i in range(len(variable_a)):
    yield variable_a[i:i+x]

variable_b=fun10(10, variable_a)
print(variable_b)
print(type(variable_b))
```

```python
#lambda and apply function
import pandas as pd
df=pd.DataFrame({'Roll_Number': [5,10,15,20] })
print(df)

df['Roll_Square']=df['Roll_Number'].apply(lambda x : x*x) #dataset e new col create kora hoyese
print(df)
print()
df
```

```python
#lambda and map function
import pandas as pd
df2=pd.DataFrame({'Roll_Number': [5,10,15,20] })
print(df2)

df2['Roll_Square']=df2['Roll_Number'].map(lambda x : x*x) #dataset e new col create kora hoyese
print(df2)
print()
df2
```

```python
#lambda and map function
import pandas as pd
df3=pd.DataFrame({'Roll_Number': [5,10,15,20] })
print(df3)

df3['Roll_grater_than_10']=df3['Roll_Number'].map(lambda x : 1 if x>10 else 0) #dataset e new col create kora hoyese
print(df3)
print()
df3
```
